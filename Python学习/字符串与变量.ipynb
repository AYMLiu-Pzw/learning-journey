## 字符串

### 介绍字符串

- 字符串本质上是文本的片段
- 字符串是字符的序列 字符是单个符号，例如字母、数字、标点符号、空格等

```python
#添加字符串
'hello'+'world'

#叫他们变成'hello world'
'hello'+' '+'world'
```

## 变量

### 介绍变量
- 一种方法来引用那些未知且可以改变的值——可以变化的值，这些被称为 变量

```python
#定义word变量
word='hello'
```

```python
#在shell单独运行会怎么样
word
```

**如果带了’ ’的变量例如’n’仅仅是’n’ 因此它被称为 字符串字面量，没有定义的变量输出会错误**

### 使用变量和 print()

```python
#定义word和your_name
word = 'Hello'
your_name='123'
```

```python
#试试'Hello ' + your_name
'Hello ' + your_name
```

```python
#把变量加在一起试试
word + your_name
#添加一个空格
word + ' ' + your_name
```

```python
#更改word的值
word = 'Goodbye'
#再次把变量加在一起试试
word + your_name
```

```python
#去掉字符串的引号试试运行
print(word + ' ' + your_name)
```
