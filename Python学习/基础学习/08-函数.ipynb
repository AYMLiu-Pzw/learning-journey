# 函数

## 定义函数

````python
# 如何编写你自己的函数
def greet(name):
    print(f"Hello {name}!")

greet("Alice")
greet("Bob")
#这定义了一个名为 greet 的函数，它接受一个参数
````

**函数定义是一个复合语句 像 if 和 for 一样，它有一个头部行，后面跟着一个缩进的主体 可以包含一个或多个语句**

````python
# 向函数添加另一个语句，使其看起来像这样：

def greet(name):
    print(f"Hello {name}!")
    print("How are you?")
# 然后再次运行程序
# 注意程序的输出是如何变化的
greet("Alice")
````

**它们让你可以多次重用相同的代码，而不必 重复自己 这就像在一个程序中编写另一个程序**
函数定义的头部总是包含以下部分：

- 特殊关键字 def，后面跟一个空格
- 函数的名称 这就像一个变量名 - 你可以选择你想要的名称，但有一些限制， 例如，它不能包含空格
- 一对括号 ( 和 )
- 括号之间零个或多个参数名称，如果有多个，用逗号分隔 在这里我们有 一个名为 name 的参数
- 一个冒号 :

## 在函数中调用函数
- 函数的主体可以包含任何内容，包括函数调用

````python
#例如，我们可以用 print_many 来实现 print_twice：  

def print_many(n, thing):
    for _ in range(n):
        print(thing)

def print_twice(x):
    print_many(2, x)

print_twice("Hello")
````

## 测试函数
- 编写程序的重要部分是测试它们是否正确工作  
你可以手动进行测试，例如通过检查 print(double(5)) 是否打印 10，但这种方式很快就会变得乏味  
- 实际上编写测试你程序的程序是很有帮助的 这被称为 自动化测试， 这些程序被称为 测试

这里有一个简单的函数 assert_equal 来帮助我们编写测试：
```
def assert_equal(actual, expected):
    if actual == expected:
        print("OK")
    else:
        print(f"Error! {repr(actual)} != {repr(expected)}")
```

````python
# 这里有一个例子供你尝试：
def assert_equal(actual, expected):
    if actual == expected:
        print("OK")
    else:
        print(f"Error! {repr(actual)} != {repr(expected)}")

def double(x):
    return x * 2

assert_equal(double(2), 4)
assert_equal(double(5), 10)
````

## return 结束函数调用
- 执行了 return 语句，函数将停止，后面的代码将被忽略
- 每次函数调用只能执行一个 return，然后执行停止

````python
# 我们更多地了解它是如何工作的
# 运行这段代码：

def foo():
    return 1
    return 2

print(foo())
````

````python
# 每次函数调用只能执行一个 return，然后执行停止
# 多个 return 语句在适当使用时仍然很有用，例如在 if-else 块中：
if condition:
    return value1
else:
    return value2
````

````python
# 一个常见的错误是误解 return 在 for 循环中的作用
# 试试以下代码：
def assert_equal(actual, expected):
    if actual == expected:
        print("OK")
    else:
        print(f"Error! {repr(actual)} != {repr(expected)}")
def double_numbers(numbers):
    for x in numbers:
        return x * 2

assert_equal(double_numbers([1, 2, 3]), [2, 4, 6])
````

````python
# 即使只有一个 return 语句，它也只会执行一次并返回一个值
# 如果你想返回多个值，可以返回一个列表：
def assert_equal(actual, expected):
    if actual == expected:
        print("OK")
    else:
        print(f"Error! {repr(actual)} != {repr(expected)}")
def double_numbers(numbers):
    doubles = []
    for x in numbers:
        doubles.append(x * 2)
    return doubles

assert_equal(double_numbers([1, 2, 3]), [2, 4, 6])
````

````python
#如果有嵌套循环会发生什么？试试以下函数：

def foo():
    for letter in 'abc':
        for number in range(3):
            print(f"{letter} {number}")
            if letter == 'b':
                return letter

foo()
````

````python
#如果将return改成break:
def foo():
    for letter in 'abc':
        for number in range(3):
            print(f"{letter} {number}")
            if letter == 'b':
                break
foo()
# 与 return 不同 
# break 只会停止它所使用的最内层循环
# 外层循环继续执行，转到下一个字母 c
# c没有触发if语句 就没有停止 他会完全执行
````
