## 列表的函数和方法

### 假设我们有一个列表 nums = [1, 2, 3]。你可以使用：

**append：**将一个元素添加到列表的末尾

nums.append(4) 将列表更改为 [1, 2, 3, 4]

**len：**返回元素的数量

len(nums) 是 3

**range：**range(n) 是一个对象，类似于从 0 到 n - 1 的数字列表 这意味着它包含 n 个数字

特别地，range(len(nums)) 就像 [0, 1, 2]，这些是 nums 中每个元素的索引

**[] (下标/索引)：**获取某个索引处的值

nums[0] 是 1，nums[1] 是 2，nums[2] 是 3

**+：**连接列表

nums + [4, 5] 是 [1, 2, 3, 4, 5]

## 字符串方法与不可变性

```python
#试试以下内容：

print('the' in 'feed the dog and the cat')
```

```python
# in 在字符串上的工作方式与在列表上相同！
# 命令返回了 True
# 因为 the 在 feed the dog and the cat 中作为 子字符串 出现
# 那么 count 和 index 呢
string = 'feed the dog and the cat'
print(string.count('the'))
print(string.index('the'))
```

```python
# 这两个方法在字符串上也像在列表上一样有效
# index 返回搜索词 'the' 
# 在更长字符串 'feed the dog and the cat' 中的 起始索引，即 5
# 列表是可变的
#  字符串是不可变的 
# - 它们没有像 append 或甚至下标赋值这样的任何方法
'Python'.append(' is cool!')
```

```python
# 由 word 引用的字符串没有被修改
# 而是 word.lower() 返回了一个新的字符串
# 但这个字符串立即被丢弃了
# 如果你想改变 word 引用的值，你必须给变量赋一个新值：
sentence = "Python rocks!"
new_sentence = sentence.upper()
print(sentence)
print(new_sentence)
# sentence.upper() 并不会改变原始的 sentence
```

## == 与 is，以及一个值的多个名称

### 是时候学习一些常常被误解并导致错误的技术细节了 运行这个程序：

list1 = [1, 2, 3]
list2 = [1, 2, 3]

print(list1)
print(list2)
print(list1 == list2)

print(list1 is list2)

list1.append(4)

print(list1)
print(list2)

```python
list1 = [1, 2, 3]
list2 = [1, 2, 3]

print(list1)
print(list2)
print(list1 == list2)

print(list1 is list2)

list1.append(4)

print(list1)
print(list2)
#创建了两个变量，它们指向列表
#两个列表有相同的元素，所以它们是相等的：list1 == list2 是 True
#但是它们是两个不同的对象，所以 list1 is list2 是 False
```

```python
#如果把“list2 = [1, 2, 3] 改为 list2 = list1”
#那么 list1 和 list2 将指向同一个列表对象，所以 list1 is list2 将是 True
list1 = [1, 2, 3]
list2 = list1

print(list1)
print(list2)
print(list1 == list2)

print(list1 is list2)

list1.append(4)

print(list1)
print(list2)
#list1 和 list2 都指向相同的列表
#list1.append(4) 会添加到那个列表 

#
```

```python
# 可以使用 copy 方法复制一个列表:
# list2=list1.copy()
list1 = [1, 2, 3]
list2=list1.copy()

print(list1)
print(list2)
print(list1 == list2)

print(list1 is list2)

list1.append(4)

print(list1)
print(list2)
# 这将使程序再次像第一个版本一样运行
```

## 在迭代时修改

```python
#它循环遍历一个数字列表，并移除小于 10 的数字
numbers = [10, 7, 8, 3, 12, 15]
for i in range(len(numbers)):
    number = numbers[i]
    if number <= 10:
        numbers.pop(i)
print(numbers)
#记住 numbers.pop(i) 会从 numbers 中移除索引为 i 的元素
#索引变量 i 按照正常的值 0, 1, 2, ... 运行
#但随着列表的变化，这些不再是我们想要的位置
#（结果会报错）
```

```python
#迭代列表时，底层仍然会遍历索引
#在迭代时绝不要修改任何东西
#保持变更和循环分开
#你可以选择在副本上循环，如下所示：
#for number in numbers.copy():
numbers = [10, 7, 8, 3, 12, 15]
for i in numbers.copy():
    if i<=10:
        numbers.remove(i)
print(numbers)
```

```python
#或在原始列表上循环并修改一个副本
numbers = [10, 7, 8, 3, 12, 15]
big_numbers=numbers.copy()

for i in numbers:
    if i<=10:
        big_numbers.remove(i)
print(big_numbers)
```

```python
#或可以从头开始构建一个新列表
numbers = [10, 7, 8, 3, 12, 15]
big_numbers=[]
for i in numbers:
    if i>10:
        big_numbers.append(i)
print(big_numbers)
```

**在迭代时绝不要修改任何东西 你的选择有：**

= **修改一个副本**

= **在副本上迭代**

- **不修改任何东西，而是创建一个新版本**
