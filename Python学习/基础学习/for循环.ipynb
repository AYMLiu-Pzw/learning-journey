# For循环

## 介绍For循环

- 它允许你重复同样的代码，Python 有两种循环：for 循环和 while 循环
- 一个循环中for/in/：/是必不可少的

```python
name = 'World'
for character in name: print(character)
```

## 缩进

- 缩进时，应该始终缩进4个空格，任何一致的缩进都是可以接受的，但4个空格是几乎所有人遵循的约定，按下tab会插入四个空格
- 缩进后会和上一段有关联，想没有关联就从开头开始

```python
name = 'World'

for character in name:
    print(character)
    print('---')
```

```python
name = 'World'

for character in name:
    print(character)
print('---')
```

## 基本术语
- 执行循环的过程称为 迭代
- 数字是不可迭代的
- 一个 表达式 是一段有值的代码。例如，在这行代码中：

sentence = 'Hello ' + name
这里有三个表达式：

'Hello '
name
'Hello ' + name
相对而言，完整的行 sentence = ... 是一个 语句。它是一个命令，告诉计算机执行一个动作,它没有自己的值

# 学后单元要点自总结

- 缩进表示执行上一条
- for _ in xx:  不管xx是变量还是'xx' 本身有几个字就会执行几次
- a=a+b 可以简化为a+=b；a=b+a 则不可简化
- 给一个空字符''定义为任何 在循环里加他数,一个for循环几次他就可以加几次数
- for没有要定义的值时通常为_ 方便阅读
