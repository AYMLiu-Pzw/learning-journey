# 布尔运算符

## 介绍or
- (或)

````python
# 想象一下你有两个朋友，Alice 和 Bob
# 下面的函数接受一个参数 name，并检查给定名称的人是否是你的朋友
def assert_equal(actual, expected):
    if actual == expected:
        print("OK")
    else:
        print(f"Error! {repr(actual)} != {repr(expected)}")
        
def is_friend(name):
    if name == "Alice":
        return True
    elif name == "Bob":
        return True
    else:
        return False

assert_equal(is_friend("Alice"), True)
assert_equal(is_friend("Bob"), True)
assert_equal(is_friend("Charlie"), False)
````


我们可以使用 or 结合 if 和 elif 语句  
or 是一个 布尔运算符， 意味着它是一个运算符（像 + 或 -）  
用于结合两个布尔值（True 或 False）  
  
表达式 A or B 如果 A 或 B 为 True  
即如果 A 为 True 或 B 为 True  
或两者都为 True，则为 True  
只有在 A 和 B 都为 False 时，它才为 False  
---(要点,在or中有一个为对 的就为对)  

将:  
```
if name == "Alice":
 return True
elif name == "Bob":
 return True
```
替换为:  
```
if name == "Alice" or name == "Bob":
 return True
```

````python
def assert_equal(actual, expected):
    if actual == expected:
        print("OK")
    else:
        print(f"Error! {repr(actual)} != {repr(expected)}")
        
def is_friend(name):
    if name == "Alice" or name == "Bob":
#通过or合成一条if
        return True
    else:
        return False

assert_equal(is_friend("Alice"), True)
assert_equal(is_friend("Bob"), True)
assert_equal(is_friend("Charlie"), False)
````

`is_friend` 只有在 `name` 是 `Alice` 或 `Bob` 时才会返回 `True`  
`or` 用于连接两个布尔值 `name == "Alice"` 和 `name == "Bob"` 之间

`name == "Alice" or name == "Bob"`本身也会返回布尔值  
所以我们可以返回布尔值 `name == "Alice" or name == "Bob"`   
可以将:  
```
if x:
    return True
else:
    return False
```
简化成:  
``
return x
``


注意我们是如何写条件的:  
``
return name == "Alice" or name == "Bob"
``  
错误是写成这样:  
``
return name == "Alice" or "Bob"
``  

`name == "Alice" or "Bob"`  
并没有翻译成  
``
name == ("Alice" or "Bob")
``  
而是:  
``
(name == "Alice") or ("Bob")
``  
当 `name == "Alice"` 为 `False` 时，它的值是 `"Bob"`

*有没有更好的方法来写条件，而不每次都重复 name ==？*  
**在 列表的函数和方法 中我们提到过 in 运算符，你可以这样与列表一起使用:**  
``
return name in ["Alice", "Bob", "Charlie"]
``

## 介绍and

另一个布尔运算符是 `and`  
表达式 `A and B` 只有在 `A` 和 `B` 都为 `True` 时才为 `True`否则就是 `False`   

#### 练习

```
def is_valid_percentage(x):
    if x < 0 or x > 100:
        return False
    else:
        return True

assert_equal(is_valid_percentage(-1), False)
assert_equal(is_valid_percentage(0), True)
assert_equal(is_valid_percentage(50), True)
assert_equal(is_valid_percentage(100), True)
assert_equal(is_valid_percentage(101), False)
```
请将这个函数重写为使用 `and`  
给定测试函数:  
```
def assert_equal(actual, expected):
    if actual == expected:
        print("OK")
    else:
        print(f"Error! {repr(actual)} != {repr(expected)}")
```

这是一个可能的解决方案：  
```
def is_valid_percentage(x):
    if 0 <= x and x <= 100:
        return True
    else:
        return False
```
和之前一样，我们可以将这个解决方案简化为：  
```
def is_valid_percentage(x):
    return 0 <= x and x <= 100
```

还有一个技巧可以进一步改进  
叫做比较链 任何这样的条件：  
``
a < b and b < c
``  
可以通过去掉多余的 and b 简化为：  
``
a < b < c
``  
这适用于任何比较运算符，包括 ==，而且这两个运算符甚至可以是不同的   
所以解决方案可以简化为：  
```
def is_valid_percentage(x):
    return 0 <= x <= 100
```

## 多行语句

代码行开始变得相当长了  
几种方法来将一个语句分散到多行中 但这不是自动的  
你要确保Python明白你正在做什么

````python
# 例如，这段代码是无效的语法，会给你一个错误：

is_friend = name == "Alice" or
            name == "Bob"
````

另一种方法是   
'确保换行符包含在某种括号内'  
这样，行 的延续是隐含的  
因为Python会等到所有括号都关闭后，才会认为一行是完整的  
如果你已经有括号了，比如你在调用一个函数  
或者创建一个列表，你可能不需要做任何事情！  
否则，你可以`在任何表达式中添加括号` 来暗示行的延续

会试图将这解释为两行独立的代码   
你需要告诉它 第一行是继续到第二行的  

一种方法是在行末添加 `\` 来“转义”换行符  

````python
name = "Bob"

is_friend = name == "Alice" or \
            name == "Bob"
# 续行符号
print(is_friend)

is_friend = (name == "Alice" or
             name == "Bob")
# 括号延续
print(is_friend)

is_friend = [name == "Alice",
             name == "Bob"]
# 有列表
print(is_friend)

print(name == "Alice" or
      name == "Bob")
# 有括号
````

## 介绍not  
- (反)

与 `and` 和 `or` 不同  
它们用于两个布尔值之间（称为 二元 运算符）   
`not` 仅在一个布尔值之前使用（称为 一元 运算符）  
**它否定所应用的表达式**  


````python
# 例如:
print(not True)
# => False
print(not False)
# => True
````

````python
# 优先值如下:
b = True
print(not b or b)
# => (not b) or b --- 而不是 not (b or b)
# => False or True 
# => True   
````

所以 如果没有括号  
`not` 的优先级高于 `or`  
`not` 的优先级也高于 `and`  
请在记不住优先级时使用**括号**或者其他的**变量**!!

#### 练习:
1. 练习：   
如果用户给你一个无法处理的文件，你想显示一个错误：    
```
if invalid_image(filename):
    print("I can't process " + filename)
```
假设 `.png` 和 `.jpg` 文件可以被处理  
但其他文件类型不能  
这里有一个示例函数来做到这一点：  
```
def invalid_image(filename):
    if filename.endswith(".png") or filename.endswith(".jpg"):
        return False
    else:
        return True

assert_equal(invalid_image("dog.png"), False)
assert_equal(invalid_image("cat.jpg"), False)
assert_equal(invalid_image("invoice.pdf"), True)
```
这比需要的要长 重写 `invalid_image`   
使得主体是一行 `return <expression>`   
即没有 `if` 语句 它应该通过相同的测试  

有两个解决方案：
```
def invalid_image(filename):
    return not (filename.endswith(".png") or filename.endswith(".jpg"))

def invalid_image(filename):
    return not filename.endswith(".png") and not filename.endswith(".jpg")
```
（这两个是等效的，因为有一个叫做德摩根定律的东西）  
可以简化的一般模式：  
如果你的代码的形式是：  
```
if x:
    return False
else:
    return True
```
其中 x 本身是一个布尔值，那么可以简化为：  
``
return not x
``

- `not`优先值比`and`高 其次是`or`
- or --- 有一个是`True`就会返回`True`
- and --- 全都是`True`才会返回`True`
- not --- 返回相反的布尔值 如`True`返回`False` 相反同理
