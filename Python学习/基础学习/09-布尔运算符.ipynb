# 布尔运算符

## 介绍or
- (或)

````python
# 想象一下你有两个朋友，Alice 和 Bob
# 下面的函数接受一个参数 name，并检查给定名称的人是否是你的朋友
def assert_equal(actual, expected):
    if actual == expected:
        print("OK")
    else:
        print(f"Error! {repr(actual)} != {repr(expected)}")
        
def is_friend(name):
    if name == "Alice":
        return True
    elif name == "Bob":
        return True
    else:
        return False

assert_equal(is_friend("Alice"), True)
assert_equal(is_friend("Bob"), True)
assert_equal(is_friend("Charlie"), False)
````


我们可以使用 or 结合 if 和 elif 语句  
or 是一个 布尔运算符， 意味着它是一个运算符（像 + 或 -）  
用于结合两个布尔值（True 或 False）  
  
表达式 A or B 如果 A 或 B 为 True  
即如果 A 为 True 或 B 为 True  
或两者都为 True，则为 True  
只有在 A 和 B 都为 False 时，它才为 False  
---(要点,在or中有一个为对 的就为对)  

将:  
```
if name == "Alice":
 return True
elif name == "Bob":
 return True
```
替换为:  
```
if name == "Alice" or name == "Bob":
 return True
```

````python
def assert_equal(actual, expected):
    if actual == expected:
        print("OK")
    else:
        print(f"Error! {repr(actual)} != {repr(expected)}")
        
def is_friend(name):
    if name == "Alice" or name == "Bob":
#通过or合成一条if
        return True
    else:
        return False

assert_equal(is_friend("Alice"), True)
assert_equal(is_friend("Bob"), True)
assert_equal(is_friend("Charlie"), False)
````

`is_friend` 只有在 `name` 是 `Alice` 或 `Bob` 时才会返回 `True`  
`or` 用于连接两个布尔值 `name == "Alice"` 和 `name == "Bob"` 之间

`name == "Alice" or name == "Bob"`本身也会返回布尔值  
所以我们可以返回布尔值 `name == "Alice" or name == "Bob"`   
可以将:  
```
if x:
    return True
else:
    return False
```
简化成:  
``
return x
``


注意我们是如何写条件的:  
``
return name == "Alice" or name == "Bob"
``  
错误是写成这样:  
``
return name == "Alice" or "Bob"
``  

`name == "Alice" or "Bob"`  
并没有翻译成  
``
name == ("Alice" or "Bob")
``  
而是:  
``
(name == "Alice") or ("Bob")
``  
当 `name == "Alice"` 为 `False` 时，它的值是 `"Bob"`

*有没有更好的方法来写条件，而不每次都重复 name ==？*  
**在 列表的函数和方法 中我们提到过 in 运算符，你可以这样与列表一起使用:**  
``
return name in ["Alice", "Bob", "Charlie"]
``

## 介绍and

另一个布尔运算符是 `and`  
表达式 `A and B` 只有在 `A` 和 `B` 都为 `True` 时才为 `True`否则就是 `False`   

#### 练习
1. 之前我们使用 `or` 写了一个函数 `is_valid_percentage`这是一个例子 解决方案：  
```
def is_valid_percentage(x):
    if x < 0 or x > 100:
        return False
    else:
        return True

assert_equal(is_valid_percentage(-1), False)
assert_equal(is_valid_percentage(0), True)
assert_equal(is_valid_percentage(50), True)
assert_equal(is_valid_percentage(100), True)
assert_equal(is_valid_percentage(101), False)
```
请将这个函数重写为使用 `and`  
给定测试函数:  
```
def assert_equal(actual, expected):
    if actual == expected:
        print("OK")
    else:
        print(f"Error! {repr(actual)} != {repr(expected)}")
```

这是一个可能的解决方案：  
```
def is_valid_percentage(x):
    if 0 <= x and x <= 100:
        return True
    else:
        return False
```
和之前一样，我们可以将这个解决方案简化为：  
```
def is_valid_percentage(x):
    return 0 <= x and x <= 100
```

还有一个技巧可以进一步改进  
叫做比较链 任何这样的条件：  
``
a < b and b < c
``  
可以通过去掉多余的 and b 简化为：  
``
a < b < c
``  
这适用于任何比较运算符，包括 ==，而且这两个运算符甚至可以是不同的   
所以解决方案可以简化为：  
```
def is_valid_percentage(x):
    return 0 <= x <= 100
```
