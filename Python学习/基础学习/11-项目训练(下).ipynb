## 使用 `input()` 的互动程序
- 我们需要一种接收玩家输入的方法--使用内置的 `input` 函数来做到这一点

````python
# 运行这个程序： 

print('Type your name, then press Enter:')
name = input()
print(f'Hello {name}!')
# 当 name = input() 运行时
# 程序实际上会停止并等待你在 shell 中输入并按 Enter
# 所以你需要输入字符后回车才能结束这个语句

#你输入的内容（不包括最后按下的 Enter）会从 input() 函数返回为字符串
````

````python
# 理解 input() 总是返回一个字符串 是至关重要的
# 无论输入了什么 你需要将这个字符串转换为你需要的类型
# 忘记这一点是导致混淆错误的常见原因

#如 这个程序一看似乎没问题 但如果你尝试一下，你会发现它实际上并不工作：

super_secret_number = 7

print("What number am I thinking of?")
guess = input()
if guess == super_secret_number:
    print("Amazing! Are you psychic?")
else:
    print("Nope!")
````

#### 练习:

使得当用户输入 `super_secret_number` 的值（在这个例子中是 7）时  
程序按预期打印 `Amazing! Are you psychic?`   
当 `super_secret_number` 是任何整数（`int`）时  
它应该可以工作  

这里有两个主要的修复方法。你可以将输入转换为数字：  
``
if int(guess) == super_secret_number:  
``  
或者将正确答案转换为字符串：  
``
if guess == str(super_secret_number):
``
这两种方法之间一个重要的区别是  
第一种方法会在用户输入不是数字的内容时引发错误

## 嵌套列表作业：在棋盘上进行移动
- 我们已经看到如何从用户那里获取输入，现在让我们用这些输入来实际在棋盘上放置棋子 并进行游戏

````python
# 试试这段代码：

def play_move(board, player):
    board[1] = player

def play_game():
    game_board = [" ", " ", " "]
    play_move(game_board, "X")
    print(game_board)

play_game()
````

注意调用 `play_move(game_board, 'X')`   
实际上是 直接修改 `game_board` 在调用 `play_move` 时的变量 `board` 和 在调用   `play_game` 时的变量 `game_board` 指向同一个列表对象  
没有复制 `Python Tutor` 很擅长用箭头来显示这一点 

这也意味着在这种情况下 `play_move` 不需要返回任何内容， 它可以直接修改 `board`，而调用者（在这种情况下是 `play_game`）将看到效果  

然而，我们的棋盘是二维的，由嵌套列表表示  
所以我们需要将 `player` 分配给内部列表的一个元素，类似于这样：  

````python
def play_move(board, player):
    row = board[1]
    row[0] = player

def play_game():
    board = [
        [" ", " ", " "],
        [" ", " ", " "],
        [" ", " ", " "],
    ]
    play_move(board, "X")
    print(board)

play_game()
````

## 制作棋盘
- 棋盘已经作为嵌套列表提供给你 但对于完整的程序，你需要自己创建它

````python
# 这里有一些代码可以做到这一点：

def make_board(size):
    row = []
    for _ in range(size):
        row.append(' ')
    board = []
    for _ in range(size):
        board.append(row)
    return board

def test():
    board = make_board(3)
    assert_equal(board, [
        [' ', ' ', ' '],
        [' ', ' ', ' '],
        [' ', ' ', ' '],
    ])
    board[0][0] = 'X'
    assert_equal(board, [
        ['X', ' ', ' '],
        [' ', ' ', ' '],
        [' ', ' ', ' '],
    ])

test()
# 这很接近，但它有一个微妙的问题
````

你能看出发生了什么吗？  

每一行的第一个位置都得到了一个 `'X'`！ 就好像代码实际上是这样做的：  
```
board[0][0] = 'X'
board[1][0] = 'X'
board[2][0] = 'X'
```
答案是:  
列表 `row` 只创建了一次，并被多次重用  
`board` 包含相同的列表三次  
不是副本，只是在三个地方的一个列表  
就像它这样做：  
``
board = [row, row, row]
``  
这意味着这段代码：  
``
board[0][0] = 'X'
``  
等同于：  
``
row[0] = 'X'  
``  
这影响了 `board` 中的“所有列表”  
因为它们都是同一个列表 `row`  
换句话说，上面的行也等同于这两行中的每一行：   
```
board[1][0] = 'X'
board[2][0] = 'X'
```
因为 `row` 同时是 `board[0]`、`board[1]` 和 `board[2]`

#### 练习:
1. 你现在的任务是修复 `make_board`  
使其没有这个问题 它仍然应该返回一个长度为 `size` 的列表  
其中每个元素也是一个长度为 `size` 的列表  
每个元素都是字符串 `' '`  
子列表应该都是独立的列表对象而  
不是同一个列表的重复  

***如下***  
这个问题可以通过将第一个循环放入第二个循环中来解决  
每次都创建一个新的 `row`：  
```
def make_board(size):
    board = []
    for _ in range(size):
        row = []
        for _ in range(size):
            row.append(' ')
        board.append(row)
    return board
```
另一种方法是每次都复制 `row` 例如保留原始代码但更改一行：  
``
board.append(row.copy())
``
你也可以用 `row[:]` 或 `list(row)` 来复制 `row` 
但重要的是要知道 所有这些方法都会对列表进行浅拷贝  
这意味着它们只复制列表的顶层 而不复制每个元素  
在这种情况下，`row` 只包含无法修改的字符串    
所以不需要复制 但如果元素是可变对象，比如列表   
就像 `board` 的情况  
你可能会再次遇到同样的问题 这里有一个例子：  
```
def make_board(size):
    row = []
    for _ in range(size):
        row.append(' ')
    board = []
    for _ in range(size):
        board.append(row.copy())
    return board

def make_cube(size):
    cube = []
    board = make_board(size)
    for _ in range(size):
        cube.append(board.copy())
    return cube

def test():
    cube = make_cube(2)
    print(cube)
    cube[0][0][0] = 'X'
    print(cube)
    print(cube[0] is cube[1])
    print(cube[0][0] is cube[0][1])
    print(cube[0][0] is cube[1][0])

test()
```  
这里 `cube` 的每个元素都是一个单独的列表  
是 `board` 的副本  
在每个副本中，每个元素也是一个单独的列表   
是 `row` 的副本   
但是 `board` 的浅拷贝都具有相同的第一个元素（第一个 `row` 的副本） 相同的第二个元素，依此类推 改变 `make_board` 不会解决任何问题  
解决方案是：  

每次调用 `make_board` 来创建一个新的 `board`，或  
使用 `deepcopy` 函数，而不是 `board.copy()`   
`deepcopy` 会在每个嵌套对象的每个层级进行拷贝
