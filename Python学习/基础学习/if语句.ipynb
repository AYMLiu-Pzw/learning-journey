## 介绍 if 语句

- **学习如何告诉计算机做出决策，并且只在某些条件下运行代码**
- **布尔值只有两个： True 和 False**

```python
#布尔值是用于 if 语句
if True:
    print('This gets printed')

if False:
    print('This does not')
```

### if 语句看起来像这样：

if <condition>:
   <body>
当计算机看到 if <condition>: 时，它会检查 <condition> 是否为 True。
如果是，它就会执行主体。 如果不是，它会跳过并继续执行程序的其余部分。

#### 练习题：
1. 修改下面的程序，添加一个额外的 布尔参数 confused，程序应该像这样开始：

sentence = 'Hello World'
excited = False
confused = True
(sentence 可以是任何字符串，两个布尔值可以是 True 或 False)

当 confused 为真时，打印的句子应该在末尾添加一个问号。 如果 confused 和 excited 都为真，句子应该以 !? 结束。

```python
#解：
sentence = 'Hello World'
excited = False
confused = True
if confused:
    sentence+='!'
if excited:
    sentence+='?'
print(sentence)
```

## 组合复合语句 
- 像 for 循环和 if 语句这样的复合语句有一个主体，它是一个内部语句的列表
- 些内部语句可以是任何东西，包括其他复合语句

```python
sentence = 'Hello World'
excited = True

if excited:
    new_sentence = ''
    for char in sentence:
        new_sentence += char + '!'
    sentence = new_sentence

print(sentence)
```

### if 和 else
- 一个 if 语句可以选择性地有一个 else 部分，**if是判断，else是表示否则的意思**如果if的条件是假，呢else就会执行

```python
#例
#字符串有各种 方法 让你方便地计算出新的值，包括 upper 和 lower。
# 这些名称指的是大写字母（大写字母）和小写字母（小写字母）
sentence = 'Hello World'
excited = True
if excited:
    sentence = sentence.upper()
else:
    sentence = sentence.lower()
print(sentence)
```

#### 练习题：
1.写一个程序，给定一个字符串 sentence，打印一个修改后的版本， 其中第一个字母大写，其余字母小写。 例如，无论输入 sentence = 'hello world' 还是 'HELLO WORLD'，输出都应该是 Hello world

```python
#解：
sentence = 'Hello World'
dui = True
a=''
for A in sentence:
    if dui:
        A=A.upper()
        dui=False
    else:
        A=A.lower()
    a+=A
print(a)
```

2. 写一个程序，以嘲讽的方式打印 sentence，例如：

OnE MoRe eXeRcIsE, aNd tHeN YoU CaN ReLaX.
每隔一个字符应该是小写，其余的应该是大写

```python
#解：
sentence = 'Hello World'
dui=True
a=''
for A in sentence:
    if dui:
        A=A.upper()
        dui=False
    else:
        A=A.lower()
        dui=True
    a+=A
print(a)
```

## 等号操作符
- 有几种方法可以获得布尔值而不直接赋值， 这使你能够构建非常有用的 if 语句，特别是有许多 比较操作符 用于比较两个表达式的值
-  最常见的是等号操作符，它检查两个值是否相等，它用两个等号表示： ==

```python
#例
print(1 + 2 == 3)
print(4 + 5 == 6)
print('ab' + 'c' == 'a' + 'bc')
```

```python
#让我们在 if 语句中使用 ==。在这个程序中，
# 只有当 c 是字符 's' 时，if 的主体才会运行

name = 'kesha'
new_name = ''
for c in name:
    if c == 's':
        c = '$'
    new_name += c

print(new_name)
```

#### 练习题：

```python
#解：
name = 'kesha'
new_name = ''
for c in name:
    if c == 's':
        c = '$'
    if c=='e':
        c='3'
    if c=='a':
        c='@'
    new_name += c

print(new_name)
```

## 介绍 elif
- 当if 代码中只需要判断一个让给予值时，可以用elif 

```python
dna = 'AGTAGCGTC'
opposite_dna = ''
for char in dna:
    if char == 'A':
        char = 'T'
    if char == 'T':
        char = 'A'
    if char == 'G':
        char = 'C'
    if char == 'C':
        char = 'G'
    opposite_dna += char

print(opposite_dna)
#T被改成了A，但A没有被改成任何东西
```

```python
# 我们可以用else来做到这一点，如下所示：
if char == 'A':
    char = 'T'
else:
    if char == 'T':
        char = 'A'
```

## 其他比较运算符
- == 的相反是 不等于 运算符 !=
-  <（小于）和 >（大于）
- 主要的比较运算符是 ==、!=、<、>、<= 和 >=

```python
#这是一个使用 != 的小程序：

sentence = 'The e key on my keyboard is broken'
new_sentence = ''
for c in sentence:
    if c != 'e':
        new_sentence += c
print(new_sentence)
```

```python
#这是一个实际的 < 运算符示例

percentage = 73

if percentage < 40:
    grade = 'F'
elif percentage < 60:
    grade = 'C'
elif percentage < 80:
    grade = 'B'
else:
    grade = 'A'

print(grade)
```
