## 介绍列表
- **列表**是Python中最常用的数据结构之一，它是一种有序的、可变的集合，可以存储任意类型的对象。列表使用方括号`[]`来定义，元素之间用逗号分隔。
- 列表是一个 序列（有序的集合/容器），可以包含任意数量的值。 这些值通常被称为 元素。 它们可以是任何东西：数字、字符串、布尔值，甚至是列表！它们也可以是不同类型的混合
- 列表是可迭代的

```python
#例：
words = ['This', 'is', 'a', 'list']

for word in words:
    print(word)
```

```python
#例：
x = 1
things = ['Hello', x, x + 3]
print(things)
```

```python
#这里有一个程序可以将列表中的所有数字相加：
numbers = [3, 1, 4, 1, 5, 9]

total = 0
for number in numbers:
    total += number

print(total)
```

### 练习题：

1. 现在修改上面程序，使其可以将字符串列表相加，而不是数字。 例如，给定：

2. 扩展程序以在每个单词 之间 插入一个分隔符字符串， 例如，给定：

words = ['This', 'is', 'a', 'list']
separator = ' - '
它将输出：

```python
#解:

words = ['This', 'is', 'a', 'list']
separator = ' - '
C=''
B=False
for A in words:
    if B:
        C+=separator
    C+=A
    B=True
print(C)
    
```

## 创建新列表
- 可以将两个列表相加，将它们组合成一个新列表
- 你也可以创建一个没有元素的空列表

```python
#例：
numbers = [1, 2] + [3, 4]
print(numbers)
new_numbers = []
new_numbers += numbers
new_numbers += [5]
print(new_numbers)
```

### 练习题：
1. 写一个程序，接受一个数字列表 并打印一个每个数字都被加倍的新列表，例如，给定：

numbers = [3, 1, 4, 1, 5, 9, 2, 6, 5]
它将打印：

[6, 2, 8, 2, 10, 18, 4, 12, 10]

```python
#解：
numbers = [3, 1, 4, 1, 5, 9, 2, 6, 5]
A=[]
for B in numbers:
    A.append(B*2)
print(A)
```

2. 现在使用 .append 写一个程序，接受一个数字列表并 打印一个只包含大于 5 的数字的新列表。

例如，给定：

numbers = [3, 1, 4, 1, 5, 9, 2, 6, 5]
它将打印：

[9, 6]
**append是将一个数加到另一个列表内，例如x=[1,2] x.append(3)就等于x=[1,2,3]**

```python
#解：
numbers = [3, 1, 4, 1, 5, 9, 2, 6, 5]
A=[]
for B in numbers:
    if B>5:
        A.append(B)
print(A)
```

## 使用 break 提前结束循环

### 练习题：
1. 编写一个程序，接受一个列表和一个值，并检查 列表是否包含该值。例如，给定：

things = ['This', 'is', 'a', 'list']
thing_to_find = 'is'
它应该打印 True，但对于

thing_to_find = 'other'
它应该打印 False

```python
#解：
things = ['This', 'is', 'a', 'list']
thing_to_find = 'is'
B=False
for A in things:
    if A==thing_to_find:
        B=True
print(B)
```

## 术语：调用函数和方法
- None 是一个特殊的“空”值,那些不想返回任何东西的函数默认返回 None,如果你看到关于 None 或 NoneType ，通常你把错误的东西赋值给了一个变量

```python
#像 len(things) 或 print(things) 这样的表达式是一个函数 调用 - 当你写下这些时，
# 你是在 调用 函数 len 或 print
# 能够这样做意味着函数是 可调用的
print(callable(len))
```

```python
#所有调用都必须返回某些东西...即使什么也不返回
# 例如，print 的工作是显示某些内容在屏幕上
# 而不是返回一个有用的值 所以它返回一些无用的东西：

things = [1, 2, 3]
length = len(things)
printed = print(length)
print(printed)
```

```python
#例如，upper 和 lower 是字符串的方法 
# 可以通过例如 word.upper() 来调用：

word = 'Hello'
print(word.upper)
print(word.upper())
```

```python
# 另一个例子是 append 是列表的方法
# 但你不能在列表上使用 .upper 或在字符串上使用 .append：
# 这会导致 AttributeError，因为这些方法不属于该类型。

word = 'Hello'
# 下面这行会导致 AttributeError，因为字符串没有 append 方法
# 如果你想在字符串末尾添加字符，可以使用 + 操作符：
# word = word + '!'
word.append('!')
```
