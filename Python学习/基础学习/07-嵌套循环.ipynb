# 嵌套循环  

## 介绍嵌套循环  

- （一个循环可以包含另一个循环）  

````python
#这有一个例子：
for letter in "ABC":
    print(letter)
    for number in range(4):
        print(f'{letter} {number}')
    print('---')
````

print(letter) 和 print('---') 各运行 3 次  
因为它们的缩进将它们放在 外部循环 中  
print(f'{letter} {number}') 被调用了 3 × 4 = 12 次  
因为它在 内部循环 for number in range(4): 中  
该循环有 4 次迭代  
但它本身在外部循环中，所以它运行了 3 次  
**这被称之为 嵌套循环**

## 介绍嵌套列表:  

````python
# 给定一个字符串列表  
# 打印列表中第二个字符串的第一个字母  
# 例如，给定：  

strings = ["abc", "def", "ghi"]  
# 你应该打印 d
# 你可能这样解决了这个问题：
string = strings[1]
print(string[0])
# 还有更简洁的方法  
# strings[1] 是一个表达式  
# 像其他任何表达式一样  
# 索引操作 [0] 可以用于任何表达式  
# 而不仅仅是变量  
# 所以你可以跳过中间变量，直接用一行代码来完成：  

print(strings[1][0])
````

````python
# 应用两次索引可以更强大  
# 我们不仅可以在字符串列表上使用它  
# 还可以在 列表的列表 上使用  
# 例如，以下程序会打印什么？

strings = [['hello', 'there'], ['how', 'are', 'you']]
print(strings[1][0])
````

**Python 允许我们使用 嵌套列表：一个每个元素都是另一个列表的列表（我们称它们为 子列表）**

## 遍历嵌套列表  
- 可以使用嵌套循环来遍历嵌套列表的每个元素和子元素  

````python
# 考虑这个嵌套列表  
numbers = [[1, 2, 3], [4, 5], [6], []]
# 输入以下嵌套循环  
for sublist in numbers:
    for num in sublist:
        print(num)
    print('---')
````
