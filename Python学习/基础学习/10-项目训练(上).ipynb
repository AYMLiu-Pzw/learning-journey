## 换行符,`format_board`
- `''.join(x)` 是一种 将指定的元素以指定的字符连接生成一个新的字符

````python
# 我们想解决显示井字棋棋盘的问题
# 这是其中一种方法：

def print_board(board):
    for row in board:
        print("".join(row))

print_board([
    ['X', 'O', 'X'],
    [' ', 'O', 'O'],
    [' ', 'X', ' ']
])
````

但理想情况下我们希望有一个  
返回字符串的函数    
而不是打印它  
我们可能想要操作这个字符串  
而在这个特定情况下我们希望能够用 assert_equal 测试它  
这是行不通的：  
``assert_equal(print_board([...]), "...")``  
因为 `print_board` 不使用 `return`  
所以它默认返回 `None`  
所以我们想要这样的代码：  
```
def format_board(board):
    ...
    return ...

assert_equal(format_board([...]), "...")
```

````python
# 然后 print(format_board(board))  
# 打印出我们在开始时看到的内容  
# 但是我们如何返回一个包含多行的字符串
# 我们如何测试它
# 我们想做这样的事情：

assert_equal(
    format_board([
        ['X', 'O', 'X'],
        [' ', 'O', 'O'],
        [' ', 'X', ' ']
    ]),
    "XOX
      OO
      X "
)
# 看看这怎么不行
````

通常字符串字面量必须在一行上  
所以这是无效的：  
```
string = "First line
Second line"
print(string)
```
使用三重引号  
(可以是 ''' 或 """) 包围字符串的内容  

````python
# 运行以下代码：

string = """First line
Second line"""
print(string)
````

三重引号字符串  
可跨越多行，且会在输出中显示  
和其他引号1一样 三重引号只是另一种表示法  
并不是一种新的字符串类型   
`"""abc"""` 和 `"abc"` 是一样的  
`\n` 代表一个 换行符 这只是另一个字符  
就像字母或空格 `(' ')` 一样  
`\n` 表示 Python 字符串字面量中的换行符  
字符串实际上并不包含 `\` 和 `n`  
**它只包含一个字符**

## 类型

各种数据类型：`字符串`、`列表`、`数字`和`布尔值`  
这些被称为 `类型` 每个值都有一个类型 它会影响它的行为   
并可以通过 `type` 函数来揭示  

````python
# type函数显示类型 
print(type('Hello World'))# =>str
print(type(23))           # =>int
print(type(True))         # =>bool
print(type([1, 2, 3]))    # =>list
print(type(4.56))         # =>float
````

`('Hello World')` 是 `<class 'str'>`  
`class` 是用来定义 类 的关键字 `str` 是字符串的缩写  

`True` 是 `bool`(布尔值缩写)  
[1, 2, 3] 的类型是 `list`  

数字有两种类型：

`int`是 '整数' 的缩写--用于表示整数，也就是说没有分数或小数点  
`float`是 '浮点数' 的缩写--用于表示带小数点的数字，可能还有分数部分  

````python
# 不同类型的值通常彼此非常不同
# 但在打印时它们可能看起来相同
# 这可能会困惑 试试这个：

print('123')
print(123)
print(123 == '123')
print('-----')
#(你可以使用 print(repr(123)) 和 print(repr('123')) 来区分它们
# repr() 函数将对象转换为供解释器读取的字符串形式

# 不同类型有不同的方法，并支持不同的运算符
# 相同的方法或运算符也可以有不同的含义
# 看 + 对于 str 和 int 的不同含义：

print(123 + 456)
print('123' + '456')
print('-----')
# 对于两个整数，+ 表示加法
# 而对于两个字符串，它表示字符串连接
# Python 会根据输入的类型自动判断 + 的含义
# 类似地，< 在两个字符串和两个整数之间的作用不同：

print(13 < 120)
print('13' < '120')
print('-----')
# 因为 13 小于 120 
# 但在两个字符串之间，它像字典排序一样工作：
# 13 在字母顺序上排在 120 之后 因为 3 在 2 之后(例如 '190' < '20')
print('190' < '20')
print('-----')

# 看看这在排序列表时有什么不同：
print(sorted([120, 13, 0]))
print(sorted(['120', '13', '0']))
# 如果你在 str 和 int 之间使用运算符，会发生什么？试试在命令行中输入：
#如 12 + '34' = TypeError: unsupported operand type(s) for +: 'int' and 'str'
````

在 `range` 中使用字符串而不是整数  
比如 `range('5')`  或者在列表下标中使用  
比如 `list['3']` 也会报错  

大多数这些问题可以通过**使用 int 函数**
将字符串转换为整数来解决：  
`int('5')` 将返回整数 `5`   
可以使用 `str` 函数将整数转换为字符串：  
`str(5)` 将返回字符串 `'5'`

2. 编写一个改进版的 `format_board` 使其具有行和列数字 如下所示：
```
 123
1XOX
2 OO
3 X 
```
它应该适用于任何单数字大小的棋盘 以下是一个测试案例：  
```
def format_board(board):
    ...

assert_equal(
    format_board([
        ['X', 'O', 'X'],
        ['O', ' ', ' '],
        [' ', 'X', 'O']
    ]),
    ' 123\n1XOX\n2O  \n3 XO'
)
```

````python
#类型 我自己解的大题
def format_board(board):
    empty_list=[]
    ah=''
    for a in range(len(board)):
        empty_a=''
        if a == 0:
            for str_H in range(len(board[0])):
                ah += str(str_H+1)
            empty_list.append(' '+str(ah))
        for A in board[a]:
            empty_a += A
        empty_list.append(str(ah[a])+empty_a)
    return "\n".join(empty_list)
            
    
    
assert_equal(
    format_board([
        ['X', 'O', 'X'],
        ['O', ' ', ' '],
        [' ', 'X', 'O']
    ]),
    ' 123\n1XOX\n2O  \n3 XO'
)
````
