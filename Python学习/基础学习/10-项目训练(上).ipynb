## 换行符,`format_board`
- `''.join(x)` 是一种 将指定的元素以指定的字符连接生成一个新的字符

````python
# 我们想解决显示井字棋棋盘的问题
# 这是其中一种方法：

def print_board(board):
    for row in board:
        print("".join(row))

print_board([
    ['X', 'O', 'X'],
    [' ', 'O', 'O'],
    [' ', 'X', ' ']
])
````

但理想情况下我们希望有一个  
返回字符串的函数    
而不是打印它  
我们可能想要操作这个字符串  
而在这个特定情况下我们希望能够用 assert_equal 测试它  
这是行不通的：  
``assert_equal(print_board([...]), "...")``  
因为 `print_board` 不使用 `return`  
所以它默认返回 `None`  
所以我们想要这样的代码：  
```
def format_board(board):
    ...
    return ...

assert_equal(format_board([...]), "...")
```

````python
# 然后 print(format_board(board))  
# 打印出我们在开始时看到的内容  
# 但是我们如何返回一个包含多行的字符串
# 我们如何测试它
# 我们想做这样的事情：

assert_equal(
    format_board([
        ['X', 'O', 'X'],
        [' ', 'O', 'O'],
        [' ', 'X', ' ']
    ]),
    "XOX
      OO
      X "
)
# 看看这怎么不行
````

通常字符串字面量必须在一行上  
所以这是无效的：  
```
string = "First line
Second line"
print(string)
```
使用三重引号  
(可以是 ''' 或 """) 包围字符串的内容  

````python
# 运行以下代码：

string = """First line
Second line"""
print(string)
````

三重引号字符串  
可跨越多行，且会在输出中显示  
和其他引号1一样 三重引号只是另一种表示法  
并不是一种新的字符串类型   
`"""abc"""` 和 `"abc"` 是一样的  
`\n` 代表一个 换行符 这只是另一个字符  
就像字母或空格 `(' ')` 一样  
`\n` 表示 Python 字符串字面量中的换行符  
字符串实际上并不包含 `\` 和 `n`  
**它只包含一个字符**

## 类型

````python
#解:
def row_winner(board):
    for a in board:
        cuo=False
        if a[0] != ' ':
            for b in a:
                if a[0] != b:
                    cuo=False
                    break
                else:
                    cuo=True
            if cuo==True:
                return cuo
    return cuo

                
            

assert_equal(
    row_winner(
        [
            ['A', 'A', 'B', 'A'],
            [' ', ' ', ' ', ' '],
            ['A', ' ', ' ', 'A'],
            ['B', ' ', 'B', 'A']
        ]
    ),
    False
)
assert_equal(
    row_winner(
        [
            ['X', ' ', 'X'],
            ['O', 'X', 'X'],
            ['O', 'O', 'O']
        ]
    ),
    True
)
````

````python
def diagonal_winner(board):
    cuo1=False
    cuo2=False
    a1=board[0][0]
    a2=board[-1][0]
    for a in range(len(board)):
        if board[a][a] == ' '\
        or board[a][a] != a1:
            cuo1 = False
            break
        else:
            cuo1 = True
    for A in range(len(board)):
        if board[-A-1][A] == ' '\
        or board[-A-1][A] != a2:
            cuo2 = False
            break
        else:
            cuo2 = True
    return (cuo1 or cuo2)
assert_equal(
    diagonal_winner(
        [
            ['O', 'X', 'O', 'X'],
            [' ', 'O', 'X', ' '],
            ['X', 'X', ' ', 'X'],
            ['X', ' ', 'O', 'O']
        ]
    ),
    True
)
assert_equal(
    diagonal_winner(
        [
            ['X', 'X', ' '],
            ['X', ' ', 'O'],
            [' ', 'O', 'O']
        ]
    ),
    False
)
````
