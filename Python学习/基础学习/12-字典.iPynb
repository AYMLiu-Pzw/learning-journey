# 字典

## 介绍字典
- 学习另一种容器类型：`dict` 即 **字典**

````python
# 你查找一个 键（例如一个单词）以获取相关的 值（例如定义或翻译）
# 这里有一个小字典，将英语单词翻译成法语
french = {'apple': 'pomme', 'box': 'boite'}
print(french['apple'])
print(french['box'])
````

`french` 是一个包含两个键值对的字典：  

`'apple': 'pomme'` 其中 `'apple'` 是键 `'pomme'` 是值    
`'box': 'boite'`   其中 `'box' `  是键`'boite'`  是值  

和列表一样 逗号` (,) `用于分隔各个项目（键值对）  
冒号`(:) `用于分隔键和值  
使用大括号 `({})` 来创建字典 而不是在编写列表时使用的方括号 `([])`  

对于列表 你是根据它们的 索引 来获取值的   
也就是它们在列表中的位置  
所以如果 `words = ['apple', 'box']`   
那么 `words[0]` 是 `'apple'，words[1]` 是 `box'`  

````python
# 现在让我们从法语翻译成英语：
french = {'apple': 'pomme', 'box': 'boite'}
print(french['pomme'])
````

你不能这样做  你只能查找一个键以获取其值，而不能反过来   
字典 `french` 只有 2 个键：`'apple'` 和 `'box'`  
`'pomme'` 是一个值，而不是一个键  

## 在实践中使用字典

````python
# 想象一下你正在构建一个在线购物网站
# 你将所有商品的价格保存在一个字典中：

prices = {'apple': 2, 'box': 5, 'cat': 100, 'dog': 100}
# 你可以看到查找字典中值的一个原因可能会出现问题
# prices[100] 会是什么？'dog'？'cat'？['dog', 'cat']？
# 另一方面，键必须是唯一的
# 想象一下，如果你的价格是这样开始的：
prices = {'apple': 2, 'apple': 3}
# 一个苹果多少钱？我们知道是 prices['apple']
# 但那是 2 还是 3？ 显然应该只有一个价格，所以不允许重复键
````

## 遍历字典键

````python
quantities = {'apple': 1, 'cat': 10}
print(quantities)
#将 print(quantities) 改为 print(quantities.keys())，并运行整个程序
````

````python
quantities = {'apple': 1, 'cat': 10}
print(quantities.keys())
````

````python
# dict 的 .keys() 可以像遍历列表一样遍历它返回的值：

quantities = {'apple': 1, 'cat': 10}
for key in quantities.keys():
    print(key)
````

````python
# 你甚至不需要 .keys()
# 直接遍历字典会自动遍历它的键
# 有时候写 .keys() 可以使你的代码更易读，但你并不是非得这样做
# 去掉 .keys() 并再次运行代码
quantities = {'apple': 1, 'cat': 10}
for key in quantities:
    print(key)
````
