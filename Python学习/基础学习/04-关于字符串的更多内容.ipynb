## 字符串中的单引号和双引号

- 如果我们想定义一个包含撇号的字符串，该怎么做？

````python
# 试试：

print('Alice's Diner')
````

````python
# 我们不能在单引号的字符串中随意使用撇号
#撇号被视为单引号，因此作为字符串定义的结束引号
#不过我们可以使用双引号替代：""
print("Alice's Diner")
````

````python
#引号只是人类表示字符串的一部分
#它们并不作为字符串的实际部分存储
#试试：
'Alice' == "Alice"
#如你所见
#Python认为用单引号或双引号定义的相同字符串是一样的
````

## f-strings

- 到目前为止，我们一直通过使用 + 来连接字符串
- 我们可以使用 f-string 更优雅地写同样的东西

````python
name = "Alice"
friend = 'Bob'
meal = "lunch"
print(name + " went to " + meal + " with " + friend + '.')

#然而，这有点繁琐。我们可以使用 f-string 更优雅地写同样的东西
#用下面的行替换上面程序的最后一行并运行它
#确保在字符串前加上 f

print(f"{name} went to {meal} with {friend}.")
````

````python
#语法以 f 开头，后面跟着一个字符串
#可以在花括号 {} 内包含变量名
#它们会被替换为那些变量的值，转换为字符串
#变量可以是任何东西：字符串、数字、列表等等

#因此，f-strings 让你轻松地组合字符串和数字
#而这些是不能 简单地相加在一起的 
#例如，运行以下程序：

name = "Alice"
age = 20
print("Hello " + name + ". You are " + age + " years old.")
````

````python
#不能在字符串和数字之间使用 +
#通过用一个单独的 f-string 替换所有的字符串连接（+）来修复这个问题

name = "Alice"
age = 20
print(f'Hello {name}. You are {age} years old.')
````

````python
#除了变量，f-string 实际上可以在花括号内包含任何 Python 表达式
print(f"2 * 3 + 4 is equal to {2 * 3 + 4}")
````
