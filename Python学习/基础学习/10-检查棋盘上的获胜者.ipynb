```python
#解:
def row_winner(board):
    for a in board:
        cuo=False
        if a[0] != ' ':
            for b in a:
                if a[0] != b:
                    cuo=False
                    break
                else:
                    cuo=True
            if cuo==True:
                return cuo
    return cuo

                
            

assert_equal(
    row_winner(
        [
            ['A', 'A', 'B', 'A'],
            [' ', ' ', ' ', ' '],
            ['A', ' ', ' ', 'A'],
            ['B', ' ', 'B', 'A']
        ]
    ),
    False
)
assert_equal(
    row_winner(
        [
            ['X', ' ', 'X'],
            ['O', 'X', 'X'],
            ['O', 'O', 'O']
        ]
    ),
    True
)
```

```python
def diagonal_winner(board):
    cuo1=False
    cuo2=False
    a1=board[0][0]
    a2=board[-1][0]
    for a in range(len(board)):
        if board[a][a] == ' '\
        or board[a][a] != a1:
            cuo1 = False
            break
        else:
            cuo1 = True
    for A in range(len(board)):
        if board[-A-1][A] == ' '\
        or board[-A-1][A] != a2:
            cuo2 = False
            break
        else:
            cuo2 = True
    return (cuo1 or cuo2)
```
